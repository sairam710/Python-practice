```python
import numpy as np

A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [5, 6],
    [7, 8]
])

print("\nAddition:")
print(A + B)

print("\nSubtraction:")
print(A - B)

print("\nElement-wise Multiplication:")
print(A * B)

print("\nMatrix Multiplication:")
print(A @ B)

print("\nTranspose of A:")
print(A.T)

print("\nInverse of A:")
print(np.linalg.inv(A))
```

```python
A = [1, 2, 3]
B = [4, 5, 6]

dot_product = 0

for i in range(len(A)):
    dot_product += A[i] * B[i]

print("Dot Product:", dot_product)
```

```python
import numpy as np

A = np.array([1, 2, 3])
B = np.array([4, 5, 6])

numpy_dot = np.dot(A, B)

print("NumPy Dot Product:", numpy_dot)
```

```python

A = [
    [1, 2],
    [3, 4]
]

B = [
    [5, 6],
    [7, 8]
]

result = [
    [0, 0],
    [0, 0]
]

for i in range(2):
    for j in range(2):
        for k in range(2):
            result[i][j] += A[i][k] * B[k][j]

print("Matrix multiplication from scratch:")

for row in result:
    print(row)
```

```python
import numpy as np

A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [5, 6],
    [7, 8]
])

numpy_result = A @ B

print("NumPy Matrix Multiplication:")
print(numpy_result)
```

```python
import numpy as np
import matplotlib.pyplot as plt

A = np.array([3, 2])
B = np.array([1, 4])
C = A + B

plt.quiver(0, 0, A[0], A[1], angles='xy', scale_units='xy', scale=1)
plt.quiver(0, 0, B[0], B[1], angles='xy', scale_units='xy', scale=1)
plt.quiver(0, 0, C[0], C[1], angles='xy', scale_units='xy', scale=1)

plt.xlim(0, 6)
plt.ylim(0, 7)
plt.grid()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

A = np.array([3, 2])
B = np.array([1, 4])

dot = np.dot(A, B)
angle = np.degrees(np.arccos(dot/(np.linalg.norm(A)*np.linalg.norm(B))))

plt.quiver(0, 0, A[0], A[1], angles='xy', scale_units='xy', scale=1)
plt.quiver(0, 0, B[0], B[1], angles='xy', scale_units='xy', scale=1)

plt.xlim(0, 5)
plt.ylim(0, 5)
plt.grid()
plt.show()

print("Dot Product:", dot)
print("Angle:", angle)
```
